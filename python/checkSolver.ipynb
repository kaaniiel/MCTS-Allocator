### Check measures from solver

```python
import json
import os
import numpy as np
import zipfile
```

```python
def check_solver_metrics(zip_file):
    with zipfile.ZipFile(zip_file, 'r') as z:
        for file in z.namelist():
            if file.endswith('.json'):
                with z.open(file) as f:
                    data = json.load(f)
                    metrics = data.get("results",{}).get("solver",{}).get("metrics", {})
                    for metric in metrics.keys():
                        if(not metrics.get("ParetoOptimal", False) and not metrics.get("EF1", False)):
                            print(f"Invalid metric in {file}: {metric}")
if __name__ == "__main__":
    # check_solver_metrics('solver_metrics.zip')
    pass
```
