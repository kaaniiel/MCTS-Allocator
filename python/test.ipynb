```python
import os
import subprocess
import json
```

```python
command = [
    "../build/Release/mcts_main.exe", 
    "-B", 
    "-t", "5", 
    "-n", "5", 
    "-o", "10", 
    "-s", "42", 
    "-r", "0,15", 
    "-MNAGJ"
]

print("Lancement du moteur C++ en cours...")

try:
    result = subprocess.run(command, capture_output=True, text=True, check=True)
    raw_output = result.stdout
    
    # ==========================================
    # L'ASTUCE DE TRONQUATURE EST ICI
    # ==========================================
    # On cherche l'index de la première accolade ouvrante
    start_idx = raw_output.find('{')
    # On cherche l'index de la dernière accolade fermante
    end_idx = raw_output.rfind('}')
    
    if start_idx != -1 and end_idx != -1:
        # On découpe la chaîne pour ne garder QUE le JSON
        json_string = raw_output[start_idx : end_idx + 1]
        
        # Maintenant on peut le parser sans erreur !
        data = json.loads(json_string)
        
        print("\n--- Succès ! Données récupérées ---")
        print(f"Meilleur score : {data['best_score']}")
        print(f"Meilleure allocation : {data['best_allocation']}")
        print(f"Itérations réalisées en 15s : {data['iterations']}")
        
    else:
        print("\n[Erreur] Aucun bloc JSON n'a été détecté dans la sortie du C++.")
        print("Sortie brute pour débug :")
        print(raw_output)

except subprocess.CalledProcessError as e:
    print(f"\n[Erreur] L'exécutable a planté avec le code {e.returncode}")
    print(f"Erreur standard (stderr) : {e.stderr}")

except json.JSONDecodeError as e:
    print(f"\n[Erreur] Le bloc extrait n'est pas un JSON valide : {e}")
    print("Bloc extrait :", json_string)
```
