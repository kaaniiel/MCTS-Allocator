```python
import json
import numpy as np
import os
import zipfile
import tqdm
import matplotlib.pyplot as plt
```

```python
def load_results(file_path):
    with open(file_path, 'r') as f:
        return json.load(f)

def load_results_from_zip(zip_path, inner_file_name):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        with zip_ref.open(inner_file_name) as f:
            return json.load(f)

```

```python
# one file
file_path = '../results/experiment_1_a5_o5_s42.json'  # Path to your JSON file
results = load_results(file_path)
print(results)
```
